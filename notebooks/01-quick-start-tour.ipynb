# Maxwell Modernized -- Quick Start Tour

A 5-minute walkthrough of the `maxwell` library. This notebook demonstrates:

1. Installation verification
2. Point charge creation and field computation
3. Lorentz force calculation
4. Speed of light relationship verification

Based on Maxwell's *A Treatise on Electricity and Magnetism* (1873).

## 1. Installation Check

Verify that maxwell is installed and check the version.

```python
import maxwell
print(f"Maxwell Modernized v{maxwell.__version__}")
print(f"Author: {maxwell.__author__}")
```

## 2. Constants and Units

Maxwell uses CGS-EMU as its primary unit system. The speed of light emerges from the electromagnetic theory.

```python
from maxwell import C, CONST

print(f"Speed of light c = {C:.4e} cm/s")
print(f"Pi = {CONST.PI:.6f}")
```

## 3. Point Charge and Electric Field

Create a point charge and compute its electric field (Art. 29-30).

```python
import numpy as np
from maxwell import PointCharge

# Create a +1 statcoulomb charge at the origin
charge = PointCharge(q=1.0, position=np.array([0.0, 0.0, 0.0]))
print(f"Charge: q = {charge.q} esu")
print(f"Position: {charge.position} cm")
```

```python
# Compute the electric field at point (1, 0, 0)
point = np.array([1.0, 0.0, 0.0])
E = charge.field_at(point)
print(f"Electric field at (1, 0, 0): E = {E} statvolt/cm")
print(f"Field magnitude: |E| = {np.linalg.norm(E):.4f}")
```

```python
# Verify inverse square law: E = q / r^2
# At r=1, q=1: E = 1/1 = 1
print(f"Expected |E| = 1.0 (for q=1, r=1)")
print(f"Computed |E| = {np.linalg.norm(E):.4f}")
```

```python
# Compute the electric potential at the same point
V = charge.potential_at(point)
print(f"Electric potential at (1, 0, 0): V = {V:.4f} statvolt")
```

## 4. Multiple Charges -- Dipole Configuration

Create an electric dipole and explore superposition.

```python
from maxwell import PointCharge

# Dipole: +q at (1, 0, 0) and -q at (-1, 0, 0)
q_plus = PointCharge(q=1.0, position=np.array([1.0, 0.0, 0.0]))
q_minus = PointCharge(q=-1.0, position=np.array([-1.0, 0.0, 0.0]))

# Field at origin (midpoint)
origin = np.array([0.0, 0.0, 0.0])
E_dipole = q_plus.field_at(origin) + q_minus.field_at(origin)
print(f"Dipole field at origin: E = {E_dipole}")
print(f"Field magnitude: |E| = {np.linalg.norm(E_dipole):.4f}")
```

## 5. Lorentz Force (Art. 490-491)

Calculate the force on a current-carrying wire in a magnetic field.

```python
from maxwell import LorentzForce

# 1 abampere wire, 10 cm long along x-axis, in 1000 gauss z-field
lf = LorentzForce(
    current=1.0,
    length=np.array([10.0, 0.0, 0.0]),
    B_field=np.array([0.0, 0.0, 1000.0]),
)
print(f"Force vector: F = {lf.force_vector} dynes")
print(f"Force magnitude: |F| = {lf.magnitude:.4f} dynes")
print(f"Force direction: {lf.direction}")
```

## 6. Speed of Light Relationship (Art. 782)

Verify that the ratio of electromagnetic to electrostatic units equals the speed of light.

```python
from maxwell.core.units import verify_speed_of_light_relationship

ratio = verify_speed_of_light_relationship()
print(f"Unit ratio / c = {ratio:.6f}")
print(f"Expected: 1.0 (the EM/ESU unit ratio equals c)")
```

## 7. Faraday Induction (Art. 528-531)

Calculate induced EMF from a changing magnetic flux.

```python
from maxwell import FaradayInduction

faraday = FaradayInduction(num_turns=100, resistance=50.0)

# Field changing at 1000 maxwells/s
emf = faraday.induced_emf(flux_change_rate=1000.0)
print(f"Induced EMF: {emf:.2f} abvolts")
print(f"Induced current: {faraday.last_induced_current:.4f} abamperes")
```

## 8. Citation System

Every function is traceable to Maxwell's original articles.

```python
from maxwell import LorentzForce
from maxwell.meta.citation import get_citation

citation = get_citation(LorentzForce.force_vector.fget)
print(f"Citation: {citation}")
```
