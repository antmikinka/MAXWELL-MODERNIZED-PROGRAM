# Maxwell Modernized -- Visualization Showcase

Gallery-style showcase of visualizations tied to specific articles in Maxwell's Treatise:

1. Method of Images (Art. 155)
2. Edge Singularities (Art. 191)
3. Hysteresis Loops (Art. 442)
4. EM Wave Propagation (Art. 791)
5. Spherical Harmonic Globes (Art. 467)
6. Unit Tubes of Flow (Art. 290)
7. Dielectric Soakage (Art. 329)
8. Magnetic Shell (Art. 409)

Each visualization is traceable to Maxwell's original text via the `@maxwell_cite` decorator.

```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# Configure inline display
%matplotlib inline

print("Visualization modules ready")
```

## 1. Method of Images -- Art. 155

A point charge above an infinite conducting plane. The image charge technique satisfies the V=0 boundary condition.

```python
from maxwell.vis import plot_method_of_images

fig, ax = plot_method_of_images(
    q=1.0, d=1.0,
    x_range=(-3.0, 3.0),
    y_range=(-3.0, 3.0),
    resolution=100,
)
plt.show()
```

## 2. Edge Singularities -- Art. 191

Power-law field enhancement near the edge of a conducting wedge. The field diverges as r^(alpha-1) near the edge.

```python
from maxwell.vis import plot_edge_singularity

fig, ax = plot_edge_singularity(
    wedge_angle=np.pi / 2,  # 90-degree wedge
    r_range=(0.01, 3.0),
    resolution=100,
)
plt.show()
```

```python
from maxwell.vis import plot_singularity_comparison

# Compare singularities for different wedge angles
fig, ax = plot_singularity_comparison(
    angles=[np.pi/4, np.pi/2, 3*np.pi/4],
    angle_labels=["45 deg", "90 deg", "135 deg"],
)
plt.show()
```

## 3. Hysteresis Loops -- Arts. 442-446

Magnetic B-H loop showing coercivity, retentivity, and energy loss per cycle.

```python
from maxwell.vis import plot_hysteresis_loops

fig, ax = plot_hysteresis_loops(
    Hc=50.0,      # Coercivity (Oe)
    Br=12000.0,   # Retentivity (gauss)
    H_max=200.0,  # Maximum applied field (Oe)
    resolution=200,
)
plt.show()
```

```python
from maxwell.vis import plot_material_comparison

# Compare hysteresis for different materials
fig, ax = plot_material_comparison()
plt.show()
```

## 4. EM Wave Propagation -- Art. 791

Orthogonal E and B fields propagating as an electromagnetic wave.

```python
from maxwell.vis import plot_em_wave_propagation

fig, ax = plot_em_wave_propagation(
    wavelength=2.0,
    amplitude=1.0,
    x_range=(0.0, 4.0),
    resolution=200,
    polarization="linear",
)
plt.show()
```

```python
from maxwell.vis import plot_wave_snapshot_3d

fig = plot_wave_snapshot_3d(
    wavelength=2.0,
    amplitude=1.0,
    x_range=(0.0, 4.0),
    resolution=100,
)
plt.show()
```

## 5. Spherical Harmonic Globes -- Art. 467

Gauss coefficient spherical harmonic decomposition of terrestrial magnetism.

```python
from maxwell.vis import plot_harmonic_globe

# l=2, m=0 mode (quadrupole)
fig = plot_harmonic_globe(l=2, m=0, resolution=50)
plt.show()
```

```python
from maxwell.vis import plot_harmonic_modes

# Compare multiple modes
fig, axes = plot_harmonic_modes(modes=[(1, 0), (2, 0), (2, 1), (3, 0)], resolution=40)
plt.show()
```

## 6. Unit Tubes of Flow -- Art. 290

Current density field visualization showing Maxwell's unit tubes of flow. Each tube carries equal flux between source and sink charges.

```python
from maxwell.vis import plot_unit_tubes_of_flow

# Default dipole: +q at (1,0,0), -q at (-1,0,0)
fig, ax = plot_unit_tubes_of_flow(
    plot_mode="streamplot",
    density=1.5,
    resolution=50,
)
plt.show()
```

```python
# Quiver mode for arrow representation
fig, ax = plot_unit_tubes_of_flow(
    plot_mode="quiver",
    density=1.0,
    resolution=40,
)
plt.show()
```

```python
# Quadrupole configuration
positions = np.array([
    [1.0, 1.0, 0.0],
    [-1.0, 1.0, 0.0],
    [1.0, -1.0, 0.0],
    [-1.0, -1.0, 0.0],
])
magnitudes = np.array([1.0, -1.0, -1.0, 1.0])

fig, ax = plot_unit_tubes_of_flow(
    charge_positions=positions,
    charge_magnitudes=magnitudes,
    x_range=(-3.0, 3.0),
    y_range=(-3.0, 3.0),
    resolution=50,
)
plt.show()
```

## 7. Dielectric Soakage -- Art. 329

Dielectric absorption (soakage) -- the slow decay of current in a charged dielectric.

```python
from maxwell.vis import plot_dielectric_soakage

# Linear scale
fig, ax = plot_dielectric_soakage(log_scale=False)
plt.show()
```

```python
# Log scale reveals the multi-time-scale decay
fig, ax = plot_dielectric_soakage(log_scale=True)
plt.show()
```

## 8. Magnetic Shell -- Art. 409

Maxwell's magnetic shell theory: the equivalence between a magnetic shell and a current loop.

```python
from maxwell.vis import plot_magnetic_shell

# 2D slice visualization
fig, ax = plot_magnetic_shell(
    loop_radius=1.0,
    x_range=(-3.0, 3.0),
    y_range=(-3.0, 3.0),
    resolution=80,
)
plt.show()
```

## 9. Field Lines and Equipotentials

Classic electrostatic field visualization.

```python
from maxwell.vis import plot_field_lines_2d

# Dipole configuration
charges = [
    {"q": 1.0, "pos": (1.0, 0.0)},
    {"q": -1.0, "pos": (-1.0, 0.0)},
]
fig, ax = plot_field_lines_2d(
    charges=charges,
    x_range=(-3.0, 3.0),
    y_range=(-3.0, 3.0),
    resolution=80,
)
plt.show()
```

```python
from maxwell.vis import plot_equipotentials_2d

# Equipotential contours for the same dipole
fig, ax = plot_equipotentials_2d(
    charges=charges,
    x_range=(-3.0, 3.0),
    y_range=(-3.0, 3.0),
    resolution=100,
    num_contours=20,
)
plt.show()
```
